```python
# W5A15
def is_right_triangle(a,b,c):
    if a>0 and b>0 and c>0:
        if a*a + b*b == c*c:
            return True
        elif a*a + c*c == b*b:
            return True
        elif b*b + c*c == a*a:
            return True
    return False
```

```python
# W5A28
def tinh_goc(x1, y1, x2, y2, x3, y3):
    import math
    ab=((x2-x1)**2+(y2-y1)**2)**0.5
    bc=((x2-x3)**2+(y2-y3)**2)**0.5
    tich_vo_huong = (x1-x2)*(x3-x2) + (y1-y2)*(y3-y2)
    cos_abc = tich_vo_huong / (ab*bc)
    return (math.acos(cos_abc) / math.pi)* 180
print(tinh_goc(1, 1, 2, 2, 4, 4))
```

```python
#W5A30
def tinh_he_so_nhi_thuc(n, k):
    n_giai_thua=1
    k_giai_thua=1
    n_k_giai_thua=1
    for i in range(1, n+1):
        n_giai_thua*=i
    for i in range(1, k+1):
        k_giai_thua*=i
    for i in range(1, n-k+1):
        n_k_giai_thua*=i 
    return int(n_giai_thua / (k_giai_thua * n_k_giai_thua))
print(tinh_he_so_nhi_thuc(7, 2))
```

```python
#W5A34
def sum_1_to_n(n:int):
    s=0
    for i in range(1,n+1):
        set_=set()
        for j in str(i):
            set_.add(j)
        if len(set_) == len(str(i)):
            s+=i
    return s
print(sum_1_to_n(12))
```

```python
# W3A4
h = int(input("Nhập chiều cao h = "))

for i in range(1, h + 1):
    if i == 1:
        # Dòng đầu tiên: chỉ in 1 dấu * ở giữa
        print(" " * (h - 1) + "*")
    elif i == h:
        # Dòng cuối cùng: in toàn bộ dấu * cho cạnh đáy
        print("*" * (2 * h - 1))
    else:
        # Các dòng ở giữa: gồm viền * 2 bên và khoảng trắng ở giữa
        khoang_trang_ngoai = " " * (h - i)
        khoang_trang_trong = " " * (2 * i - 3)
        print(khoang_trang_ngoai + "*" + khoang_trang_trong + "*")
```

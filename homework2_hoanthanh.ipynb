```python
# Bài 3
import math
a = float(input("Nhập cạnh a: "))
b = float(input("Nhập cạnh b: "))
c = float(input("Nhập cạnh c: "))

if a + b > c and a + c > b and b + c > a:
    chu_vi = a + b + c
    p = chu_vi / 2
    dien_tich = math.sqrt(p * (p - a) * (p - b) * (p - c))
    
    if a == b == c:
        loai = "Tam giác đều"
    elif a == b or a == c or b == c:
        if math.isclose(a**2 + b**2, c**2) or math.isclose(a**2 + c**2, b**2) or math.isclose(b**2 + c**2, a**2):
            loai = "Tam giác vuông cân"
        else:
            loai = "Tam giác cân"
    elif math.isclose(a**2 + b**2, c**2) or math.isclose(a**2 + c**2, b**2) or math.isclose(b**2 + c**2, a**2):
        loai = "Tam giác vuông"
    else:
        loai = "Tam giác thường"
    
    print(f"Đây là {loai}")
    print(f"Chu vi: {chu_vi}, Diện tích: {dien_tich}")
else:
    print("Khong la ba canh cua mot tam giac")
```

```python
# Bài 5
import math
a = float(input("Nhập a: "))
b = float(input("Nhập b: "))
c = float(input("Nhập c: "))

if a == 0:
    if b == 0:
        print("Phương trình vô số nghiệm" if c == 0 else "Phương trình vô nghiệm")
    else:
        print(f"Nghiệm x = {-c/b}")
else:
    delta = b**2 - 4*a*c
    if delta < 0:
        print("Phương trình vô nghiệm")
    elif delta == 0:
        print(f"Phương trình có nghiệm kép x1 = x2 = {-b/(2*a)}")
    else:
        x1 = (-b + math.sqrt(delta)) / (2*a)
        x2 = (-b - math.sqrt(delta)) / (2*a)
        print(f"Phương trình có 2 nghiệm phân biệt: x1 = {x1}, x2 = {x2}")
```

```python
# Bài 8
a = float(input("Nhập a: "))
b = float(input("Nhập b: "))
m = float(input("Nhập m: "))
c = float(input("Nhập c: "))
d = float(input("Nhập d: "))
n = float(input("Nhập n: "))

D = a * d - b * c
Dx = m * d - b * n
Dy = a * n - m * c

if D == 0:
    if Dx == 0 and Dy == 0:
        print("Hệ phương trình vô số nghiệm")
    else:
        print("Hệ phương trình vô nghiệm")
else:
    x = Dx / D
    y = Dy / D
    print(f"Hệ phương trình có nghiệm duy nhất: x = {x}, y = {y}")
```

```python
# Bài 10
import math

x_c = float(input("Nhập hoành độ tâm đường tròn: "))
y_c = float(input("Nhập tung độ tâm đường tròn: "))
r = float(input("Nhập bán kính đường tròn: "))

x_a = float(input("Nhập hoành độ điểm A: "))
y_a = float(input("Nhập tung độ điểm A: "))

d = math.sqrt((x_a - x_c)**2 + (y_a - y_c)**2)

if math.isclose(d, r):
    print("Điểm A nằm trên đường tròn")
elif d < r:
    print("Điểm A nằm trong đường tròn")
else:
    print("Điểm A nằm ngoài đường tròn")
```

```python
#W4A6
a, b= map(int,input().split())
s=0
for i in range(a,b+1):
    for j in range(2,int(i**0.5)+1) :
        if i%j ==0:
            break
    else :
        s+=i
    if i==1:
        s=s-1
print(s)
```

```python
# W4A7
n= int(input())
lst=[]
for i in range(1,n+1):
    if n%i==0 :
        for j in range(2,int(i**0.5)+1) :
            if i%j==0 :
                break
        else :
            lst.append(i)
print(lst[len(lst)-1])
```

```python
# W4A10
n=int(input())
so_tot_nhat= 1
do_dai_lon_nhat= 1
for i in range(1,n+1):
    copy=i
    copy_len= 1
    while i!=1 :
        if i%2==0 :
            i=i/2
        else :
            i= i*3 +1
        copy_len += 1
    if copy_len > do_dai_lon_nhat:
        do_dai_lon_nhat = copy_len
        so_tot_nhat = copy
print(so_tot_nhat, do_dai_lon_nhat)
```

```python
# W4A13
a=int(input())
b=int(input())
Sa=0
Sb=0
for i in range(1,a):
    if a%i==0 :
        Sa+=i
for i in range(1,b):
    if b%i==0 :
        Sb+=i
if Sa==b and Sb==a :
    print("True")
else :
    print("False")
```
